```python
X = df[df.origin == 'Asia'][['vehicle_weight', 'model_year']].head(n=7).values
```

```python
X
```

```python
XTX = X.T.dot(X)
```

```python
XTX
```

```python
XTX_inv = np.linalg.inv(XTX)
```

```python
XTX_inv
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```

```python
w = XTX_inv.dot(X.T).dot(y)
```
